```python

import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the' | 'a'
N -> 'cat' | 'dog' | 'mouse'
V -> 'chased' | 'saw'
""")

parser = nltk.TopDownChartParser(grammar)

sentence = "the cat chased a mouse".split()

print("Parse Trees:")
for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```

```python

import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the' | 'a'
N -> 'boy' | 'girl' | 'ball'
V -> 'kicked' | 'saw'
""")

parser = nltk.EarleyChartParser(grammar)

sentence = "the boy kicked a ball".split()

print("Earley Parse Results:")
for tree in parser.parse(sentence):
    print(tree)
```

```python


import nltk

grammar = nltk.CFG.fromstring("""
S -> SG_NP SG_VP | PL_NP PL_VP

SG_NP -> 'the cat' | 'the dog'
PL_NP -> 'the cats' | 'the dogs'

SG_VP -> 'runs' | 'sleeps'
PL_VP -> 'run' | 'sleep'
""")

parser = nltk.ChartParser(grammar)

sentences = [
    "the cat runs",
    "the cats run",
    "the cat run",
    "the dogs sleeps"
]

for sentence in sentences:
    tokens = sentence.split()

    try:
        valid = any(parser.parse(tokens))
    except ValueError:
        valid = False

    print(sentence, "->", "Correct" if valid else "Incorrect")
```

```python

import nltk

grammar = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.6] | N [0.4]
VP -> V NP [0.7] | V [0.3]
Det -> 'the' [0.5] | 'a' [0.5]
N -> 'boy' [0.5] | 'dog' [0.5]
V -> 'sees' [1.0]
""")

parser = nltk.ViterbiParser(grammar)

sentence = "the boy sees a dog".split()

print("Probabilistic Parse:")
for tree in parser.parse(sentence):
    print(tree)
    print("Probability:", tree.prob())
```

```python


import re

TOKEN = re.compile(
    r"\s*(forall|exists|[A-Za-z_]\w*|->|[~&|(),.])"
)

def tokenize(text):
    tokens = []
    pos = 0

    while pos < len(text):
        if text[pos:].strip() == "":
            break

        match = TOKEN.match(text, pos)
        if not match:
            raise ValueError("Invalid token")

        tokens.append(match.group(1))
        pos = match.end()

    return tokens


def parse(text):
    tokens = tokenize(text)
    index = [0]

    def peek():
        if index[0] < len(tokens):
            return tokens[index[0]]
        return None

    def take(expected=None):
        token = peek()

        if token is None or (
            expected is not None and token != expected
        ):
            raise ValueError("Unexpected token")

        index[0] += 1
        return token

    def primary():
        if peek() == "~":
            take("~")
            primary()

        elif peek() in ("forall", "exists"):
            take()

            variable = peek()
            if variable is None or not re.fullmatch(
                r"[A-Za-z_]\w*", variable
            ):
                raise ValueError("Expected variable")

            take()

            if peek() == ".":
                take(".")

            primary()

        elif peek() == "(":
            take("(")
            expression()
            take(")")

        else:
            predicate = take()

            if not re.fullmatch(r"[A-Za-z_]\w*", predicate):
                raise ValueError("Expected predicate")

            take("(")

            # Read at least one argument
            arg = take()
            if not re.fullmatch(r"[A-Za-z_]\w*", arg):
                raise ValueError("Expected argument")

            while peek() == ",":
                take(",")
                arg = take()
                if not re.fullmatch(r"[A-Za-z_]\w*", arg):
                    raise ValueError("Expected argument")

            take(")")

    def expression():
        primary()

        while peek() in ("&", "|", "->"):
            take()
            primary()

    expression()

    if peek() is not None:
        raise ValueError("Extra tokens")

    return "Valid FOPC expression"


expressions = [
    "Human(x)",
    "forall x. Human(x)",
    "Human(x) & Mortal(x)",
    "Human(x) -> Mortal(x)",
    "Human(x) &"
]

for expr in expressions:
    try:
        print(expr, ":", parse(expr))
    except (ValueError, IndexError):
        print(expr, ": Invalid FOPC expression")
```

```python

import nltk
from nltk.wsd import lesk

nltk.download("wordnet")
nltk.download("omw-1.4")

sentence = "I deposited money in the bank"

words = sentence.split()

# Find the meaning of "bank" from context
sense = lesk(words, "bank", pos="n")

if sense:
    print("Word: bank")
    print("Synset:", sense.name())
    print("Meaning:", sense.definition())
else:
    print("No sense found")
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Python is a programming language",
    "Python is used for machine learning",
    "Natural language processing uses machine learning",
    "Information retrieval searches documents"
]

query = "Python machine learning"

# Convert documents into TF-IDF vectors
vectorizer = TfidfVectorizer()
doc_vectors = vectorizer.fit_transform(documents)

# Convert query into a TF-IDF vector
query_vector = vectorizer.transform([query])

# Calculate cosine similarity
scores = cosine_similarity(query_vector, doc_vectors)[0]

# Rank documents by relevance
ranked = sorted(
    enumerate(scores, start=1),
    key=lambda item: item[1],
    reverse=True
)

print("Document Ranking:")
for doc_id, score in ranked:
    print(f"Document {doc_id}: Score = {score:.4f}")
    print(documents[doc_id - 1])
```

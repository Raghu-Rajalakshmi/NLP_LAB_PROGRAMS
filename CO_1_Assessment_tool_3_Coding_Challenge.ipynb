```python

n, k = map(int, input().split())
alphabet = input().split()
start = int(input())

data = list(map(int, input().split()))
m = data[0]
accepting = set(data[1:1 + m])

t = int(input())

transitions = {}

for _ in range(t):
    p, c, r = input().split()
    p = int(p)
    r = int(r)

    if (p, c) not in transitions:
        transitions[(p, c)] = set()

    transitions[(p, c)].add(r)

Q = int(input())

for _ in range(Q):
    s = input().strip()

    if s == "#":
        s = ""

    current = {start}

    for c in s:
        next_states = set()

        for state in current:
            next_states.update(transitions.get((state, c), set()))

        current = next_states

        if not current:
            break

    if current & accepting:
        print("ACCEPT")
    else:
        print("REJECT")
```

```python

def vowel(c):
    return c in "aeiou"


def doublable(word):
    if len(word) < 3 or word[-1] in "wxy":
        return False

    if vowel(word[-3]) or not vowel(word[-2]) or vowel(word[-1]):
        return False

    groups = 0
    in_group = False

    for c in word:
        if vowel(c):
            if not in_group:
                groups += 1
                in_group = True
        else:
            in_group = False

    return groups == 1


def generate(stem, feature):
    if feature in ("PL", "3SG"):
        if stem.endswith(("s", "x", "z", "ch", "sh")):
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("o") and not vowel(stem[-2]):
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("y") and not vowel(stem[-2]):
            return stem[:-1] + "ies"
        return stem + "s"

    if feature == "PROG":
        if (len(stem) >= 3 and stem.endswith("e")
                and not vowel(stem[-2])):
            return stem[:-1] + "ing"
        if doublable(stem):
            return stem + stem[-1] + "ing"
        return stem + "ing"

    suffix = {"PAST": "ed", "COMP": "er", "SUP": "est"}[feature]

    if stem.endswith("e"):
        return stem + suffix[1:]
    if len(stem) >= 2 and stem.endswith("y") and not vowel(stem[-2]):
        return stem[:-1] + "i" + suffix
    if doublable(stem):
        return stem + stem[-1] + suffix
    return stem + suffix


S = int(input())
lexicon = []

for _ in range(S):
    stem, pos = input().split()
    lexicon.append((stem, pos))

Q = int(input())

features = {
    "N": ["PL"],
    "V": ["3SG", "PROG", "PAST"],
    "ADJ": ["COMP", "SUP"]
}

for _ in range(Q):
    surface = input().strip()
    analyses = []

    for stem, pos in lexicon:
        if surface == stem:
            analyses.append(stem + "+" + pos)

        for feature in features[pos]:
            if generate(stem, feature) == surface:
                analyses.append(stem + "+" + pos + "+" + feature)

    analyses = sorted(set(analyses))

    if analyses:
        print(", ".join(analyses))
    else:
        print("UNKNOWN")
```

```python

def is_consonant(word, i):
    c = word[i]

    if c in "aeiou":
        return False

    if c == 'y':
        if i == 0:
            return True
        return not is_consonant(word, i - 1)

    return True


def measure(word):
    m = 0
    prev_vowel = False

    for i in range(len(word)):
        curr_vowel = not is_consonant(word, i)

        if prev_vowel and not curr_vowel:
            pass

        if i > 0:
            prev_is_consonant = is_consonant(word, i - 1)
            curr_is_consonant = is_consonant(word, i)

            if prev_is_consonant and not curr_is_consonant:
                j = i
                while j < len(word) and not is_consonant(word, j):
                    j += 1
                if j < len(word):
                    m += 1

        prev_vowel = curr_vowel

    return m


def has_vowel(word):
    for i in range(len(word)):
        if not is_consonant(word, i):
            return True
    return False


def ends_double_consonant(word):
    return (
        len(word) >= 2
        and word[-1] == word[-2]
        and is_consonant(word, len(word) - 1)
    )


def cvc(word):
    if len(word) < 3:
        return False

    n = len(word)
    return (
        is_consonant(word, n - 3)
        and not is_consonant(word, n - 2)
        and is_consonant(word, n - 1)
        and word[-1] not in "wxy"
    )


def porter_step1(word):
    # Step 1a
    if word.endswith("sses"):
        word = word[:-2]
    elif word.endswith("ies"):
        word = word[:-2]
    elif word.endswith("ss"):
        pass
    elif word.endswith("s"):
        word = word[:-1]

    # Step 1b
    removed = False

    if word.endswith("eed"):
        stem = word[:-3]
        if measure(stem) > 0:
            word = stem + "ee"
    elif word.endswith("ed") and has_vowel(word[:-2]):
        word = word[:-2]
        removed = True
    elif word.endswith("ing") and has_vowel(word[:-3]):
        word = word[:-3]
        removed = True

    # Cleanup after ed or ing removal
    if removed:
        if word.endswith(("at", "bl", "iz")):
            word += "e"
        elif ends_double_consonant(word) and word[-1] not in "lsz":
            word = word[:-1]
        elif measure(word) == 1 and cvc(word):
            word += "e"

    # Step 1c
    if word.endswith("y") and has_vowel(word[:-1]):
        word = word[:-1] + "i"

    return word


Q = int(input())

for _ in range(Q):
    word = input().strip()
    print(porter_step1(word))
```
